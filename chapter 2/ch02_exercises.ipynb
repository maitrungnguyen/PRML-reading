Recall the Bernoulli distribution:
\begin{align*}
P(X = x \mid p) = p^x (1-p)^{1-x}
\end{align*}
where $x \in \{0, 1\}$. Then we have:
\begin{align*}
&\sum_{x \in \{0, 1\}} P(X = x \mid p) = P(X = 0 \mid p) + P(X = 1 \mid p) = (1-p) + p = 1\\
&E[X] = 0 \cdot P(X = 0 \mid p) + 1 \cdot P(X = 1 \mid p) = 0 \cdot (1-p) + 1 \cdot p = p\\
&\text{Var}(X) = E[X^2] - (E[X])^2 = E[X] - (E[X])^2 = p - p^2 = p(1-p)
\end{align*}
The entropy of the Bernoulli distribution can be calculated as follows:
\begin{align*}
H(X) &= -\sum_{x \in \{0, 1\}} P(X = x \mid p) \log P(X = x \mid p)\\
&= -P(X = 0 \mid p) \log P(X = 0 \mid p) - P(X = 1 \mid p) \log P(X = 1 \mid p)\\
&= -(1-p) \log (1-p) - p \log p
\end{align*}


We have:
\begin{align*}
\Gamma(a)\Gamma(b) &= \int_0^\infty \exp(-x) x^{a-1} dx \cdot \int_0^\infty \exp(-y) y^{b-1} dy\\
&= \int_0^\infty \int_0^\infty \exp(-(x+y)) x^{a-1} y^{b-1} dx dy
\end{align*}
Now we can make a change of variables. Let $t = x + y$ with $x$ is fixed. Then we have $y = t - x$ and $dy = dt$. The limits of integration for $t$ will be from $x$ to $\infty$. Thus, we can rewrite the integral as:
\begin{align*}
\Gamma(a)\Gamma(b) &= \int_0^\infty \int_x^\infty \exp(-t) x^{a-1} (t-x)^{b-1} dt dx\\
&= \int_0^\infty \exp(-t) \left( \int_0^t x^{a-1} (t-x)^{b-1} dx \right) dt
\end{align*}

Then let $x = \mu t$. Then we have $dx = t d\mu$. The limits of integration for $\mu$ will be from $0$ to $1$. we can rewrite the integral as:
\begin{align*}
\Gamma(a)\Gamma(b) &= \int_0^\infty \exp(-t) \left( \int_0^1 (\mu t)^{a-1} (t - \mu t)^{b-1} t d\mu \right) dt\\
&= \int_0^\infty \exp(-t) t^{a+b-1} \left( \int_0^1 \mu^{a-1} (1-\mu)^{b-1} d\mu \right) dt\\
&= \int_0^\infty \exp(-t) t^{a+b-1} dt \cdot \int_0^1 \mu^{a-1} (1-\mu)^{b-1} d\mu\\
&= \Gamma(a+b) \cdot \int_0^1 \mu^{a-1} (1-\mu)^{b-1} d\mu
\end{align*}

Hence, we have:
\begin{align*}
\int_0^1 \mu^{a-1} (1-\mu)^{b-1} d\mu = \frac{\Gamma(a)\Gamma(b)}{\Gamma(a+b)}
\end{align*}




The Gaussian distribution of $x_1,x_2$ with precion $\rho_1$ and $\rho_2$:
\begin{align*}
p(x_1) &= \sqrt{\frac{\rho_1}{2\pi}} \exp\left(-\frac{\rho_1}{2} (x_1 - \mu_1)^2\right)\\
p(x_2) &= \sqrt{\frac{\rho_2}{2\pi}} \exp\left(-\frac{\rho_2}{2} (x_2 - \mu_2)^2\right)
\end{align*}

With $x = x_1 + x_2$, we have:
\begin{align*}
p(x) &= \int_{-\infty}^\infty p(x \mid x_2) p(x_2) dx_2\\
&= \int_{-\infty}^\infty p(x_1 = x - x_2) p(x_2) dx_2\\
\end{align*}

Consider the exponent of the integrand:
\begin{align*}
&-\frac{\rho_1}{2} (x - x_2 - \mu_1)^2 - \frac{\rho_2}{2} (x_2 - \mu_2)^2\\
\end{align*}

Omit the constant term, goup terms involving $x_2$ together, we have:
\begin{align*}
(\rho_1 + \rho_2) x_2^2 - 2(\rho_1 (x - \mu_1) + \rho_2 \mu_2) x_2 + (\rho_1 (x - \mu_1)^2 + \rho_2 \mu_2^2)
\end{align*}

Complete the square, we have:
\begin{align*}
&(\rho_1 + \rho_2) \left(x_2 - \frac{\rho_1 (x - \mu_1) + \rho_2 \mu_2}{\rho_1 + \rho_2}\right)^2 + \frac{\rho_1 \rho_2}{\rho_1 + \rho_2} (x - \mu_1 - \mu_2)^2
\end{align*}

Hence, we have:
\begin{align*}
p(x) &= \int_{-\infty}^\infty \sqrt{\frac{\rho_1}{2\pi}} \sqrt{\frac{\rho_2}{2\pi}} \exp\left(-\frac{1}{2} (\rho_1 + \rho_2) \left(x_2 - \frac{\rho_1 (x - \mu_1) + \rho_2 \mu_2}{\rho_1 + \rho_2}\right)^2 - \frac{1}{2} \frac{\rho_1 \rho_2}{\rho_1 + \rho_2} (x - \mu_1 - \mu_2)^2\right) dx_2\\
\propto& \exp\left(-\frac{1}{2} \frac{\rho_1 \rho_2}{\rho_1 + \rho_2} (x - \mu_1 - \mu_2)^2\right)
\end{align*}

Then $x$ is distributed according to a Gaussian distribution with mean $\mu_1 + \mu_2$ and precision $\frac{\rho_1 \rho_2}{\rho_1 + \rho_2}$.

The diffential entropy of a Gaussian distribution with mean $\mu$ and precision $\rho$ can be calculated as follows:
\begin{align*}
H(X) &= -\int_{-\infty}^\infty p(x) \log p(x) dx\\
&= -\int_{-\infty}^\infty \sqrt{\frac{\rho}{2\pi}} \exp\left(-\frac{\rho}{2} (x - \mu)^2\right) \log \left(\sqrt{\frac{\rho}{2\pi}} \exp\left(-\frac{\rho}{2} (x - \mu)^2\right)\right) dx\\
&= -\int_{-\infty}^\infty \sqrt{\frac{\rho}{2\pi}} \exp\left(-\frac{\rho}{2} (x - \mu)^2\right) \left(\frac{1}{2} \log \frac{\rho}{2\pi} - \frac{\rho}{2} (x - \mu)^2\right) dx\\
&= -\frac{1}{2} \log \frac{\rho}{2\pi} + \frac{\rho}{2} \int_{-\infty}^\infty \sqrt{\frac{\rho}{2\pi}} \exp\left(-\frac{\rho}{2} (x - \mu)^2\right) (x - \mu)^2 dx\\
&= -\frac{1}{2} \log \frac{\rho}{2\pi} + \frac{1}{2}
\end{align*}

Then the differential entropy of $x$ is:
\begin{align*}
H(X) &= -\frac{1}{2} \log \frac{\rho_1 \rho_2}{2\pi (\rho_1 + \rho_2)} + \frac{1}{2}\\

We have $D$ eigenvectors for a $D \times D$ symmetric matrix $\mathbf{\Sigma}$. The identity presentation:
\begin{align*}
\mathbf{I} = \sum_{i=1}^D \mathbf{u}_i \mathbf{u}_i^T
\end{align*}
where $\mathbf{u}_i$ is the $i$-th eigenvector. Then we can express the symmetric matrix $\mathbf{\Sigma}$ as:
\begin{align*}
\mathbf{\Sigma} = \mathbf{\Sigma} \mathbf{I} = \mathbf{\Sigma} \sum_{i=1}^D \mathbf{u}_i \mathbf{u}_i^T = \sum_{i=1}^D \mathbf{\Sigma} \mathbf{u}_i \mathbf{u}_i^T = \sum_{i=1}^D \lambda_i \mathbf{u}_i \mathbf{u}_i^T
\end{align*}

Consider $S = \sum_{i=1}^D \frac{1}{\lambda_i} \mathbf{u}_i \mathbf{u}_i^T$. Then we have:
\begin{align*}
\mathbf{\Sigma} S &= \mathbf{\Sigma} \sum_{i=1}^D \frac{1}{\lambda_i} \mathbf{u}_i \mathbf{u}_i^T = \sum_{i=1}^D \frac{1}{\lambda_i} \mathbf{\Sigma} \mathbf{u}_i \mathbf{u}_i^T = \sum_{i=1}^D \frac{1}{\lambda_i} \lambda_i \mathbf{u}_i \mathbf{u}_i^T = \sum_{i=1}^D \mathbf{u}_i \mathbf{u}_i^T = \mathbf{I}
\end{align*}

Hence, we have $\mathbf{\Sigma}^{-1} = \sum_{i=1}^D \frac{1}{\lambda_i} \mathbf{u}_i \mathbf{u}_i^T$.

```python
import numpy as np

def manual_multivariate_normal(mean, cov):
    K = np.linalg.cholesky(cov) # create matrix K such that cov = K @ K.T
    d = len(mean)
    z = np.random.normal(size=d) # sample from standard normal
    return mean + K @ z

# X = [manual_multivariate_normal(mean=[0, 0], cov=[[1, 5], [5, 1]]) for i in range(1000)]
X = [manual_multivariate_normal(mean=[0, 0], cov=[[6, 2], [2, 9]]) for i in range(1000)]
sample_var = np.var(X, axis=0)
print("Sample variance:", sample_var)
print("Std:", np.sqrt(sample_var))
```

Recall the t-distribution:
\begin{align*}
P(X = x \mid \nu) = \frac{\Gamma\left(\frac{\nu + 1}{2}\right)}{\sqrt{\nu \pi} \Gamma\left(\frac{\nu}{2}\right)} \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu + 1}{2}}
\end{align*}

When $\nu \to \infty$, consider the $x$ dependent part:
\begin{align*}
\lim_{\nu \to \infty} \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu + 1}{2}} &= \lim_{\nu \to \infty} \exp\left(-\frac{\nu + 1}{2} \log\left(1 + \frac{x^2}{\nu}\right)\right)\\
&= \lim_{\nu \to \infty} \exp\left(-\frac{\nu + 1}{2} \cdot \frac{x^2}{\nu}\right)\\
&= \exp\left(-\frac{x^2}{2}\right)
\end{align*}

We observe that when $\nu \to \infty$, the t-distribution $P(X = x \mid \nu) \propto \exp\left(-\frac{x^2}{2}\right)$, which is the probability density function of the standard normal distribution.



Likelihood function for histogram-like density model:
\begin{align*}
L(\mathbf{p} \mid \mathbf{x}) = \prod_{n=1}^N h_{k_n}^{n_{k_n}}\\
\propto \log L(\mathbf{p} \mid \mathbf{x}) = \sum_{n=1}^N \log h_{k_n} = \sum_{k=1}^K N_k \log h_k
\end{align*}

Normolization constraint:
\begin{align*}
\sum_{k=1}^K h_k \Delta_k = 1 
\end{align*}

Lagrange-multiplier:
\begin{align*}
\mathcal{L}(\mathbf{h}, \lambda) = \sum_{k=1}^K N_k \log h_k + \lambda \left(1 - \sum_{k=1}^K h_k \Delta_k\right)
\end{align*}

Set the derivative of $\mathcal{L}$ with respect to $h_k$ to zero, we have:
\begin{align*}
\frac{\partial \mathcal{L}}{\partial h_k} = \frac{N_k}{h_k} - \lambda \Delta_k = 0
\end{align*}

Hence, we have:
\begin{align*}
h_k = \frac{N_k}{\lambda \Delta_k}
\end{align*}

Plug $h_k$ into the normalization constraint, we have:
\begin{align*}
\sum_{k=1}^K \frac{N_k}{\lambda \Delta_k} \Delta_k = 1 \Rightarrow \lambda = N
\end{align*}

Therefore, the maximum likelihood solution is:
\begin{align*}
h_k = \frac{N_k}{N \Delta_k}
\end{align*}
